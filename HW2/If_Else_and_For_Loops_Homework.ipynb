### Exercise 9: FizzBuzz (4 points)
Process integers 1 through 30, inclusive, using a loop. Print exactly one line per integer:
- `FizzBuzz` if divisible by both 3 and 5.
- `Fizz` if divisible by 3 only.
- `Buzz` if divisible by 5 only.
- Otherwise, print the integer.

**Check:** 3 → `Fizz`; 5 → `Buzz`; 15 and 30 → `FizzBuzz`.

*Hint: check the combined condition first.*

```python
for number in range(1, 31):
    if number % 3 == 0 and number % 5 == 0:
        print("FizzBuzz")
    elif number % 3 == 0:
        print("Fizz")
    elif number % 5 == 0:
        print("Buzz")
    else:
        print(number)
```
